# Dataset 1: Pakistani E-Commerce Data 

```python
import pandas as pd
```

```python
import matplotlib.pyplot as plt
```

```python
import seaborn as sns
```

```python
df = pd.read_csv("pakistani_ecommerce_dataset.csv")
```

```python
df
```

```python
df['product_category'].value_counts()
```

```python
df['order_value'].mean().round(2)
```

```python
df['city'].value_counts()
```

```python
df['payment_method'].value_counts()
```

```python
city_average = df.groupby('city')['order_value'].mean()

city_average.plot(kind='bar', figsize=(8,5))

plt.title('Average Order Value by City')
plt.xlabel('City')
plt.ylabel('Average Order Value (Rs.)')
plt.xticks(rotation=45)
plt.show()
```
