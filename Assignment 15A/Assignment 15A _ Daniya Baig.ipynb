```python
import seaborn as sns
```

```python
df = pd.read_csv("ride_hailing_dataset.csv")
```

```python
df.head()
```

```python
df.groupby('service_type')['fare'].mean().round(2)
```

```python
average_fare = df.groupby('service_type')['fare'].mean()

average_fare.plot(kind='bar', figsize=(8,5))

plt.title('Average Fare by Service Type')
plt.xlabel('Service Type')
plt.ylabel('Average Fare (Rs.)')

plt.xticks(rotation=0)
plt.show()
```
